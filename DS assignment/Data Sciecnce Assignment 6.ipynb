# Part A: Hypothesis Testing

```python
import math
from scipy.stats import norm
```

```python
# Given values
x_bar = 3050
mu = 4000
sigma = 125
n = 25
alpha = 0.05
```

```python
# Z-test calculation
# Calculate the Test Statistic:
z = (x_bar - mu) / (sigma / math.sqrt(n))
```

```python
# Critical value
z_critical = norm.ppf(1 - alpha)
```

```python
#Determine the Probability and compare

print("Z value:", z)
print("Critical Z value:", z_critical)

if z > z_critical:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

### Conclusion

#### At the 5% significance level, the sample data does not provide sufficient evidence that the actual weekly operating cost is higher than the theoretical cost model.

# Part B: Confidence Intervel

#### a. Build 99% Confidence Interval Using Sample Standard Deviation

```python
import numpy as np
from scipy.stats import t

data = np.array([
    1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32,
    0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29
])

n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)

alpha = 0.01
t_critical = t.ppf(1 - alpha/2, df=n-1)

margin_error = t_critical * (std / np.sqrt(n))

ci_lower = mean - margin_error
ci_upper = mean + margin_error

print("99% CI (Unknown σ):", (ci_lower, ci_upper))
```

#### b. Build 99% Confidence Interval Using Known Population Standard Deviation

```python
from scipy.stats import norm

sigma = 0.2
z_critical = norm.ppf(1 - alpha/2)

margin_error = z_critical * (sigma / np.sqrt(n))

ci_lower = mean - margin_error
ci_upper = mean + margin_error

print("99% CI (Known σ):", (ci_lower, ci_upper))
```
