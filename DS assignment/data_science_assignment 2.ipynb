### 1) Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.


```python
my_list=[28,12,23,25,44,46,98]
print(my_list)
```

```python
# Append operation
my_list.append(50)
print(my_list)       # value add the end of list
```

```python
# Extend operation 
list=[1,2,3,4,5]
my_list.extend(list)  # add multiple items
print(my_list)
```

```python
# Insert operation
my_list.insert(1,29)
print(my_list)  # add the value at specific position
```

```python
# remove operation
my_list.remove(98)
print(my_list)        # Removing values
```

```python
# pop operation
my_list.pop()
print(my_list)  # Removing last value
```

```python
my_list.clear()
my_list   # removing all values in the list
```

```python
my_list
```

```python
# index opeartion
print(my_list.index(44))   # Find the index number of specific values
```

```python
# Count operation
print(my_list.count(44))  # count that how many times is occured
```

```python
# Sort Function
my_list.sort()
my_list         # Arrange the data in ascending order
```

```python
my_list.sort(reverse=True)
my_list                 # Arrange the data in descending order
```

### 6) Given a 'marks' variable, print the grade:
### - 90 and above: 'A'
### - 75–89: 'B'
### - 50–74: 'C'
### - Below 50: 'Fail'


```python
# We use elif conditional statement for checking multiple condition

marks=int(input("enter your marks"))
if marks>=90 and marks<=100:
    print("A Grade")
elif marks>=75 and marks<=89:
    print("B Grade")
elif marks>=50 and marks<=74:
    print("C Grade")
else:
    print("fail")
```
