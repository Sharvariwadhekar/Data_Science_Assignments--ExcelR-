# Dimensionality Reduction with PCA

```python
## Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

#### Apply PCA

```python
from sklearn.decomposition import PCA

pca = PCA()
pca_data = pca.fit_transform(scaled_data)
```

```python
plt.figure(figsize=(8,5))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Scree Plot")
plt.grid()
plt.show()
```

```python
## PCA Transformation
pca_final = PCA(n_components=2)
pca_transformed = pca_final.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_transformed, columns=["PC1", "PC2"])
pca_df.head()
```

# Clustering with Original Data

```python
# Apply K-Means
```

```python
from sklearn.cluster import KMeans

kmeans_original = KMeans(n_clusters=3, random_state=42)
labels_original = kmeans_original.fit_predict(scaled_data)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(scaled_data[:,0], scaled_data[:,1], c=labels_original, cmap='viridis')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clustering (Original Data)")
plt.show()
```

```python
# Evaluation Metrics
from sklearn.metrics import silhouette_score, davies_bouldin_score

sil_original = silhouette_score(scaled_data, labels_original)
db_original = davies_bouldin_score(scaled_data, labels_original)

print("Silhouette Score (Original):", sil_original)
print("Davies-Bouldin Index (Original):", db_original)
```

# Clustering with PCA Data

```python
kmeans_pca = KMeans(n_clusters=3, random_state=42)
labels_pca = kmeans_pca.fit_predict(pca_df)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_df["PC1"], pca_df["PC2"], c=labels_pca, cmap='viridis')
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering (PCA Data)")
plt.show()
```

```python
## Evaluation Metrics
sil_pca = silhouette_score(pca_df, labels_pca)
db_pca = davies_bouldin_score(pca_df, labels_pca)

print("Silhouette Score (PCA):", sil_pca)
print("Davies-Bouldin Index (PCA):", db_pca)
```

 # Comparison & Analysis

```python
comparison = pd.DataFrame({
    "Method": ["Original Data", "PCA Data"],
    "Silhouette Score": [sil_original, sil_pca],
    "Davies-Bouldin Index": [db_original, db_pca]
})

comparison
```

### PCA improves clustering clarity
###  PCA reduces noise and redundancy
###  Slight loss of information is acceptable

# Conclusion & Insights

#### Wine dataset contains highly correlated features

#### PCA successfully reduced dimensions while retaining ~95% variance
