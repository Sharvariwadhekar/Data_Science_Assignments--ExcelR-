```python
###2
### only billing relevent columns
billing_cols=['PatientID','Department','Doctor','BillAmount']

patient_billing_df = patient_df[billing_cols] 
```

```python
###3
### drop administrative columns
patient_df = patients_df.drop(
    columns=['ReceptionistID','CheckInTime'],
    errors='ignore'
)
```

```python
###4
### total bill amount per department
department_revenue=(
    patient_billing_df
    .groupby('Department')['BillAmount']
    .sum()
    .reset_index()
)
print(department_revenue)
```

```python
### 5
### remove duplicate patient records
patient_df = patient_df.drop_duplicates(subset='PatientID',keep='last')
```
