```python
# Separate Features and Target
X = df.iloc[:, :-1] 
y = df.iloc[:, -1]    
```

```python
# Encode Target Variable
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y) 
```

```python
# Feature Scaling (Very Important for ANN)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded, test_size=0.2, random_state=42
)
```

# Model Implementation (Basic ANN)

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([
    Dense(32, activation='relu', input_shape=(60,)),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train the Model
history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
# Model Evaluation (Baseline)
test_loss, test_accuracy = model.evaluate(X_test, y_test)
test_accuracy
```

# Evaluation Metrics

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

y_pred = (model.predict(X_test) > 0.5).astype(int)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))

print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

# Hyperparameter Tuning

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import accuracy_score
import pandas as pd
```

```python
def build_model(neurons=32, learning_rate=0.001):
    model = Sequential([
        Dense(neurons, activation='relu', input_shape=(60,)),
        Dense(neurons // 2, activation='relu'),
        Dense(1, activation='sigmoid')
    ])

    optimizer = Adam(learning_rate=learning_rate)

    model.compile(
        optimizer=optimizer,
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
results = []

neurons_list = [16, 32, 64]
learning_rates = [0.001, 0.01]
batch_sizes = [8, 16]
epochs = 50

for neurons in neurons_list:
    for lr in learning_rates:
        for batch in batch_sizes:
            model = build_model(neurons=neurons, learning_rate=lr)

            model.fit(X_train, y_train,
                      epochs=epochs,
                      batch_size=batch,
                      verbose=0)

            y_pred = (model.predict(X_test) > 0.5).astype(int)
            acc = accuracy_score(y_test, y_pred)

            results.append({
                'neurons': neurons,
                'learning_rate': lr,
                'batch_size': batch,
                'accuracy': acc
            })
```

```python
results_df = pd.DataFrame(results)
results_df.sort_values(by='accuracy', ascending=False).head()
```
